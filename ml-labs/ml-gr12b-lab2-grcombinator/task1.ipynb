```python
price_per_loaf = 3.52        
discount_rate = 0.55         
quantity = int(input("Введите количество приобретенных вчерашних буханок хлеба: "))

discounted_price = price_per_loaf * (1 - discount_rate) 
total_cost = discounted_price * quantity                 

print(f"Обычная цена за буханку:    {price_per_loaf:>8.2f} руб.")
print(f"Цена со скидкой:            {discounted_price:>8.2f} руб.")
print(f"Общая стоимость:            {total_cost:>8.2f} руб.")
```
