```python
a = int(input("Enter level of noise"))
if a > 130:
    print("Too high")
elif a < 40:
    print("Too low")
else:
    if a == 130:
        print("Отбойный молоток")
    elif a == 106:
        print("Газовая газонокосилка")
    elif a == 70:
        print("Будильник")
    elif a == 40:
        print("Тихая комната")
    elif a < 130 and a > 106:
        print("между газонокосилкой и молотком")
    elif a < 106 and a > 70:
        print("между будильником и газовой газонокосилкой")
    else:
        print("Между тихой комнаой и будильником")
```
