```python
import numpy as np

c = 3e8

def distance(p1, p2):
    return np.sqrt((p1[0]-p2[0])**2 + (p1[1]-p2[1])**2)

def estimate_position(receivers, tdoa, c):
    grid_x = np.linspace(0, 1000, 50)
    grid_y = np.linspace(0, 1000, 50)
    best_pos = None
    best_error = float('inf')

    for x in grid_x:
        for y in grid_y:
            pos = np.array([x, y])
            dist = np.array([distance(pos, r) for r in receivers])
            est_tdoa = (dist[1:] - dist[0]) / c
            error = np.sum((est_tdoa - tdoa)**2)
            if error < best_error:
                best_error = error
                best_pos = pos
    return best_pos

def read_point(index):
    x = float(input(f"Введите координату X приемника {index}: "))
    y = float(input(f"Введите координату Y приемника {index}: "))
    return np.array([x, y])

def read_time(index):
    t = float(input(f"Введите время прибытия сигнала на приемник {index} (в секундах): "))
    return t

print("Введите координаты трех приемников:")
receivers = np.array([read_point(i) for i in range(1, 4)])

print("Введите время прибытия сигнала на каждый приемник (первый приемник - эталон):")
times = np.array([read_time(i) for i in range(1, 4)])

tdoa = times[1:] - times[0]

position = estimate_position(receivers, tdoa, c)
print(f"\nОценённые координаты объекта: x = {position[0]:.1f} м, y = {position[1]:.1f} м")
```
