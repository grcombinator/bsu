```python
from itertools import permutations

def solve(n, mink, maxk):
    digits = str(n)
    possible_numbers = set()
    
    for length in range(1, len(digits) + 1):
        for perm in permutations(digits, length):
            s = ''.join(perm)
            if s[0] == '0' and len(s) > 1:
                continue
            num = int(s)
            if num > 0:
                possible_numbers.add(num)

                

    n_set = set()
    for num in possible_numbers:
        if 2 * num >=mink and 2 * num <= maxk:
            n_set.add(2*num)
        if num%2 == 0 and num/2 >=mink and num/2<=maxk:
            n_set.add(int(num/2))

    n_list = list(n_set)
    n_list.sort()
    return n_list

n = int(input())
mink = int(input())
maxk = int(input())
print(solve(n, mink, maxk))

    
```
