```python
import panel as pn
import pandas as pd
import plotly.express as px

# Настройка Panel
pn.extension("plotly")

# Создаём тестовые данные
df = pd.DataFrame({
    'x': range(10),
    'y': [i**2 for i in range(10)]
})

# Создаём график
fig = px.line(df, x='x', y='y', title="Пример графика")

# Создаём виджеты
slider = pn.widgets.IntSlider(name='Число точек', start=1, end=10, value=5)

# Функция обновления графика
@pn.depends(slider.param.value)
def update_plot(n):
    df_filtered = df.head(n)
    return px.line(df_filtered, x='x', y='y', title=f"График с {n} точками")

# Собираем интерфейс
app = pn.Column(
    "# Panel App: Интерактивный график",
    slider,
    update_plot
)

# Отображаем приложение в ноутбуке
app
```

```python
# Добавляем вкладки
tab1 = pn.Column("Таблица данных", df.head())
tab2 = pn.Column("График", fig)

tabs = pn.Tabs(("Данные", tab1), ("График", tab2))

# Добавляем тему
template = pn.template.FastListTemplate(
    title="Мой Panel App",
    sidebar=[slider],
    main=[tabs],
    accent_base_color="#FF6B6B",
    header_background="#2C3E50"
)

template.servable()
```
